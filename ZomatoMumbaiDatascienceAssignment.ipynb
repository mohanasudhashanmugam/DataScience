```python
##Cleaning dataset
```

```python
##a.removing redundant rows
```

```python
##wrong_data=raw_df[raw_df["PAGE NO"]=="PAGE NO"]
```

```python
#raw_df=raw_df[~wrong_data]
raw_df = raw_df[raw_df["PAGE NO"] != "PAGE NO"]
```

```python
#dropping columns which are not needed for analysis
raw_df.drop(['URL','PAGE NO','CITY'],axis=1,inplace=True)
```

```python
raw_df.head()
```

```python
##b.REMOVING NULL RECORDS
```

```python
raw_df.isnull().sum()
```

```python
raw_df[raw_df['PRICE'].isnull()]
```

```python
## dropping the above row from dataset

raw_df=raw_df.drop(labels=15080,axis=0)
```

```python
## replacing other null records with NA
raw_df.fillna("NA", inplace=True)
```

```python
raw_df.isnull().sum()
```

```python
raw_df.info()
```

```python
#converting numerical columns to numeric datatype
```

```python
# Checking for text values in the column before converting it to numeric datatype

##raw_df["PRICE"].value_counts()

raw_df[pd.to_numeric(raw_df["PRICE"], errors='coerce').isna()]
```

```python
#This confirms column has only numeric values
```

```python
# Checking for text values in the column before converting it to numeric datatype

##raw_df["Rating"].value_counts()

raw_df[pd.to_numeric(raw_df["RATING"], errors='coerce').isna()]
```

```python
raw_df["RATING"].value_counts()
```

```python
#replacing text values with 0
raw_df["RATING"].replace(to_replace=['-','Opening','NEW'],value=0,inplace=True)
```

```python
# Checking for text values in the column before converting it to numeric datatype

##raw_df["VOTES"].value_counts()

raw_df[pd.to_numeric(raw_df["VOTES"], errors='coerce').isna()]
```

```python
raw_df["VOTES"].value_counts()
```

```python
#replacing text values with 0
raw_df["VOTES"].replace(to_replace=['-','Opening','NEW'],value=0,inplace=True)
```

```python
#changing the datatype of numeric value column to numeric
```

```python
raw_df["VOTES"]=raw_df["VOTES"].astype('int64')
raw_df["PRICE"]=raw_df["PRICE"].astype('int64')
raw_df["RATING"]=raw_df["RATING"].astype('float64')
```

```python
raw_df.info()
```

```python
#working with TIMING column
```

```python
raw_df["TIMING"].value_counts()
```

```python
#Splitting the TIMING column and storing it in different columns
```

```python
temp_df=raw_df["TIMING"].str.split("(",n=1,expand=True)
temp_df
```

```python
##Assigning the columns back to raw_df
raw_df["TIMING"]=temp_df[0]
raw_df['DAYS_OPEN']=temp_df[1]
raw_df.head()
```

```python
raw_df["DAYS_OPEN"]=raw_df["DAYS_OPEN"].str.replace(")","",regex=False)
```
