```python
#3.Answer
import numpy as np
from scipy.stats import zscore

data = [1, 2, 3, 100, 5, 6, 7, 8, 9]

z_scores= zscore(data)

outliers = [i for i, z in zip(data, z_scores) if z>2]
print(outliers)
```

```python
import numpy as np
from scipy.stats import zscore

data = [1, 2, 3, 100, 5, 6, 7, 8, 9]
median = np.median(data)

data = [median if abs(zscore(x)) > 2 else x for x in data]
```

```python
##4.Answer
import numpy as np
from scipy.stats import zscore

data = [1, 2, 3, 100, 5, 6, 7, 8, 9]
median = np.median(data)
z_score=abs(zscore(data))
data = [median if z > 2 else x for x,z in zip(data,z_score)]
data
```
