```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
```

```python
dataset=pd.read_csv("Placement_Data_Full_Class.csv")
dataset.drop(['sl_no','salary'],axis=1,inplace=True) ###to drop multiple colmn use ['columns' ]
dataset['status'].value_counts()
```

```python
dataset=pd.get_dummies(dataset,drop_first=True)
dataset
##dataset['salary'].fillna(0,inplace=True)  ##Replacing Null values with constant '0'
```

```python
independent=dataset.drop(['status_Placed'],axis=1)

X=independent

X = sm.add_constant(X)
```

```python
dependent=dataset[['status_Placed']]
y=dependent
```

```python
logit_model = sm.Logit(y, X)
result = logit_model.fit()
```

```python
print(result.summary())
```

```python
print(result.params)
```

```python
print(result.pvalues)
```

```python
print(result.conf_int())
```

```python
odds_ratios = np.exp(result.params)

print(odds_ratios)
```

```python
summary_df = pd.DataFrame({
    'Coefficient': result.params,
    'P-value': result.pvalues,
    'Odds Ratio': np.exp(result.params)
})

print(summary_df)
```
