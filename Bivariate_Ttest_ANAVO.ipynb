```python
import pandas as pd
dataset=pd.read_csv('Placement_Data_Full_Class.csv')
dataset
```

```python
## Find similarity between Commerce and science group student's mba_p marks
## we use Independant - Unpaired - Ttest - to find similarity between two different groups with same condition
## p_value range should be less 5% or 0.05, for high similarities
```

```python
dataset.dropna(inplace=True)
dataset
```

```python
from scipy.stats import ttest_ind
Commerce = dataset[dataset["hsc_s"]=='Commerce']["mba_p"]
Science = dataset[dataset["hsc_s"]=='Science']["mba_p"]
ttest_ind(Commerce,Science)
```

```python
## In the result, p_value is 21% which is not in acceptable range, so there are more differences in the 
## mba_p marks between Commerce and Science groups students of hsc
```

```python
## Find similarity between Commerce group student's mba_p and etest_p marks
## we use dependant - paired - Ttest - to find similarity with in same group with different conditions
## p_value range should be less 5% or 0.05, for high similarities
```

```python
from scipy.stats import ttest_rel
Commerce = dataset[dataset["hsc_s"]=='Commerce']["etest_p"]
Commerce1 = dataset[dataset["hsc_s"]=='Commerce']["mba_p"]
ttest_rel(Commerce,Commerce1)
```

```python
## In the result, p_value is 662% which is not in acceptable range, so the differences between
## mba_p and etest_p marks of Commerce group students of hsc is way higher
```

```python
## H0- Null Hypothesis- There is no significant difference between columns 'ssc_p','hsc_p','degree_p'
## H1- Alternate Hypothesis- There are significant difference between columns 'ssc_p','hsc_p','degree_p'
```

```python
import pandas as pd
dataset=pd.read_csv('Placement_Data_Full_Class.csv')
```

```python
from scipy.stats import stats
stats.f_oneway(dataset['ssc_p'],dataset['hsc_p'],dataset['degree_p'])
```

```python
Here, we see the pvalue is not <= 0.05,so H0 Null hypothesis is rejected and H1 Alternate hypothesis is accepted.
Hence there are significant difference between columns 'ssc_p','hsc_p','degree_p'
```

```python
## calculate two way classification
```

```python
##Gender:
## H0- Null Hypothesis- There is no significant differences in ssc_p marks due to 'gender'
## H1- Alternate Hypothesis- There are significant differences in ssc_p marks due to 'gender'

##ssc_b
## H0- Null Hypothesis- There is no significant differences in ssc_p marks due to 'ssc_b'
## H1- Alternate Hypothesis- There are significant differences in ssc_p marks due to'ssc_b'
```

```python
import statsmodels.api as sm
from statsmodels.formula.api import ols
df = dataset
model = ols('ssc_p ~ C(gender) * C(ssc_b)', data=df).fit()
anova_table = sm.stats.anova_lm(model, typ=2)

print(anova_table)
```

```python
## gender:
## Since P-value 0.297036 > 0.05, we accept H1. So there are significant differences in ssc_p marks due to 'gender'

##ssc_b
## Since P-value 0.085980 > 0.05, we accept H1. So there are significant differences in ssc_p marks due to'ssc_b'
```
